```python
#5
a=[int(x) for x in input().split()]
dct={
    'positives':0,'negatives':0,'zeros':0
     }
for i in a:
    if i==0:
        dct["zeros"]+=1
    elif i>0:
        dct["positives"]+=1
    else:
        dct["negatives"]+=1
print(dct)
```

```python
#9
d1 = eval(input())
d2 = eval(input())
d = {}
for k in d1:
    d[k] = d1[k]
for k in d2:
    if k in d:
        d[k] += d2[k]
    else:
        d[k] = d2[k]
for k in sorted(d):
    print(f"{k}: {d[k]}")
```

```python
#10
a=[int(x) for x in input().split()]
k=int(input())
kq=[]
for i in range(0,len(a)):
    for j in range(i,len(a)):
        if a[i]+a[j]==k and i!=j:
            if (a[i],a[j]) not in kq:
                kq.append((a[i],a[j]))
kq.sort()
print(kq)

```
