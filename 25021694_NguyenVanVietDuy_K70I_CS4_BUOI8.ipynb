```python
#1
R=3.14
class hinhtru:
    def __init__(self,r,h):
        self.r=r
        self.h=h
    def smat(self):
        return 2*R*self.r*self.h + 2*R*self.r*self.r

    def v(self):
        return R * self.r * self.r * self.h

r,h=map(int,input().split())
kq=hinhtru(r,h)
print(kq.smat())
print(kq.v())

```

```python
#2
class date:
    def __init__(self,d,m,y):
        self.d=d
        self.m=m
        self.y=y
    def nhuan(self):
        return self.y % 400 == 0 or (self.y % 4 == 0 and self.y % 100 != 0)
    def check(self):
        if self.m==2:
            if self.nhuan():
                if self.d>29:
                    return False
            else:
                if self.d>28:
                    return False
        if self.d>30 and self.m in [4,6,9,11]:
            return False
        if self.d>31 or self.d<1 or self.m<1:
            return False
        return True
    def tang(self):
        dx=0
        mx=0
        yx=0
        if self.m in [4,6,9,11] and self.d==30:
            dx=1
            mx=self.m+1
        if self.m not in [2,4,6,9,11] and self.d==31:
            dx=1
            mx=self.m+1
        if self.m==2:
            if self.d==29 and self.nhuan():
                dx=1
                mx=3
                yx=self.y+1
            if self.d==28 and not self.nhuan():
                dx=1
                mx=3
                yx=self.y+1
        if mx>12:
            mx=mx%12
            yx=self.y+1
        if dx==0 and mx==0 and yx==0:
            dx=self.d+1
            mx=self.m
            yx=self.y
        return str(dx) +' '+ str(mx) +' '+ str(yx)
d,m,y=map(int,input().split())
da=date(d,m,y)
if da.check()==True:
    print(da.tang())
else:
    print("INVALID")
```

```python
#3
class point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def ans(self):
        if self.x==0==self.y:
            return "goc TD"
        elif self.x==0:
            return "nam tren truc hoanh"
        elif self.y==0:
            return "nam tren truc tung"
        else:
            return "diem binh thuong"
    def dist(self):
        return (self.x**2+self.y**2)**0.5
x,y=map(int,input().split())
diem=point(x,y)
print(diem.ans())
print(diem.dist())

```

```python
#5
class giohang:
    def __init__(self,a):
        self.a=a

    def empty(self):
        self.a=[]
        return "da xoa"

    def them(self,b):
        self.a.append(b)
        return self.a

    def xoa(self, ten):
        for i in range(len(self.a)):
            if self.a[i][0] == ten:
                self.a.pop(i)
                return "Da xoa san pham"
        return "Khong tim thay"

    def checkempty(self):
        return len(self.a)==0

    def hienthi(self):
        return self.a

    def tinhtien(self):
        t=0
        for i in self.a:
            t+=i[1]
        return t
c="NO"
a=[]
mt=giohang(a)
index = """
Chon tac vu:
2.them
3.xoa
4.kiem tra rong
5.tinh tong tien
6.in het ra
7.xoa toan bo
"""

while c!="YES":
    k=int(input(index))

    if k==2:
        ten = input("Nhap ten: ")
        gia = int(input("Nhap gia: "))
        print(mt.them([ten, gia]))

    elif k==3:
        ten = input("Nhap ten can xoa: ")
        print(mt.xoa(ten))

    elif k==4:
        print(mt.checkempty())

    elif k==5:
        print(mt.tinhtien())

    elif k==6:
        print(mt.hienthi())

    elif k==7:
        print(mt.empty())

    c=input("Ban co muon thoat khong?: ")

```

```python
#7
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return abs(a)

class psx:
    def __init__(self, tu1, mau1, tu2, mau2):
        g1 = gcd(tu1, mau1)
        g2 = gcd(tu2, mau2)
        self.tu1 = tu1 // g1
        self.mau1 = mau1 // g1
        self.tu2 = tu2 // g2
        self.mau2 = mau2 // g2

    def nhan(self):
        tu = self.tu1 * self.tu2
        mau = self.mau1 * self.mau2
        g = gcd(tu, mau)
        return f"{tu//g}/{mau//g}"

    def chia(self):
        tu = self.tu1 * self.mau2
        mau = self.mau1 * self.tu2
        g = gcd(tu, mau)
        return f"{tu//g}/{mau//g}"

    def cong(self):
        bcnn = self.mau1 * self.mau2 // gcd(self.mau1, self.mau2)
        tu = self.tu1 * (bcnn // self.mau1) + self.tu2 * (bcnn // self.mau2)
        g = gcd(tu, bcnn)
        return f"{tu//g}/{bcnn//g}"

    def tru(self):
        bcnn = self.mau1 * self.mau2 // gcd(self.mau1, self.mau2)
        tu = self.tu1 * (bcnn // self.mau1) - self.tu2 * (bcnn // self.mau2)
        g = gcd(tu, bcnn)
        return f"{tu//g}/{bcnn//g}"

tu1,mau1,op,tu2,mau2=map(str,input().split())
ps=psx(int(tu1),int(mau1),int(tu2),int(mau2))
if op=="+":
    print(ps.cong())
elif op=='-':
    print(ps.tru())
elif op=='*':
    print(ps.nhan())
elif op=='/':
    print(ps.chia())


```

```python
#9
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(score)

    def avg(self):
        return sum(self.scores) / len(self.scores)

    def rank(self):
        a = self.avg()
        if a >= 8:
            return "Excellent"
        elif a >= 6.5:
            return "Good"
        elif a >= 5:
            return "Average"
        else:
            return "Poor"

name = input()
n = int(input())

st = Student(name)

for _ in range(n):
    subject, score = input().split()
    st.add_score(subject, float(score))

avg_score = st.avg()
print(f"{st.name} {avg_score:.2f} {st.rank()}")

```

```python
#10
class Book:
    def __init__(self, t, a, y):
        self.t = t
        self.a = a
        self.y = y

class Library:
    def __init__(self):
        self.l = []

    def add(self, book):
        self.l.append(book)

    def counta(self, name):
        d = 0
        for i in self.l:
            if i.a == name:
                d += 1
        return d

    def county(self, yr):
        d = 0
        for i in self.l:
            if i.y == yr:
                d += 1
        return d

t = int(input())
l = Library()

for _ in range(t):
    s = input()
    a = s.split()

    if a[0] == "ADD":
        a2 = a[1].split(";")
        l.add(Book(a2[0], a2[1], int(a2[2])))

    elif a[0] == "COUNT":
        print(l.counta(a[1]))

    elif a[0] == "COUNTYEAR":
        print(l.county(int(a[1])))

```

```python
#11
import math
def taovector(P, Q):
    return (Q[0] - P[0], Q[1] - P[1], Q[2] - P[2])
def tichcohuong(u, v):
    return (
        u[1]*v[2] - u[2]*v[1],
        u[2]*v[0] - u[0]*v[2],
        u[0]*v[1] - u[1]*v[0])
def tichvohuong(u, v):
    return u[0]*v[0] + u[1]*v[1] + u[2]*v[2]
def dodai(u):
    return math.sqrt(tichvohuong(u, u))
A = list(map(float, input().split()))
B = list(map(float, input().split()))
C = list(map(float, input().split()))
D = list(map(float, input().split()))
phaptuyen1 = tichcohuong(taovector(A, B), taovector(A, C))
phaptuyen2 = tichcohuong(taovector(B, C), taovector(B, D))
cosphi = abs(tichvohuong(phaptuyen1, phaptuyen2)) / (
    dodai(phaptuyen1) * dodai(phaptuyen2))
cosphi = max(-1.0, min(1.0, cosphi))
phi = math.degrees(math.acos(cosphi))
print(f"{phi:.2f}")

```

```python
#12
class sp:
    def __init__(self,t,a,t1,a1):
        self.t=t
        self.a=a
        self.t1=t1
        self.a1=a1
    def tich(self):
        return [self.t*self.t1-self.a*self.a1,self.t*self.a1+self.a*self.t1]
t,a,t1,a1=map(int,input().split())
st=sp(t,a,t1,a1)
for i in st.tich():
    print(i,end=' ')
```

```python
#13
def checktg(a,b,c):
    if a+b<=c or b+c<=a or a+c<=b:
        return False
    else:
        return True
class sp:
    def __init__(self,t,a,t1):
        self.t=t
        self.a=a
        self.t1=t1
    def s(self):
        if checktg(self.t,self.a,self.t1)==False:
            return -1
        else:
            p=(self.t+self.a+self.t1)/2
            return (p*(p-self.t)*(p-self.a)*(p-self.t1))**0.5
t,a,t1=map(int,input().split())
st=sp(t,a,t1)
print(st.s())
```
