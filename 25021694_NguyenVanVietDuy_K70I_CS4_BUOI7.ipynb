```python
#1
def br(a,left,right,n):
    while left<=right:
        mid=(right+left)//2
        if a[mid]<n:
            left = mid + 1
        elif a[mid]>n:
            right= mid - 1
        elif a[mid]==n:
            return mid
    return -1

a=[int(x) for x in input().split()]
n=int(input())
a.sort()
print(br(a,0,len(a)-1,n))

```

```python
#2
def count(a,n):
    d=0
    for i in a:
        if i==n:
            d+=1
    return d
a=[int(x) for x in input().split()]
n=int(input())
print(count(a,n))
```

```python
#3
def bbsort(a):
    for i in range(0,len(a)-1):
        for j in range(i+1,len(a)):
            if a[i]>a[j]:
                k=a[i]
                a[i]=a[j]
                a[j]=k
    return a
a=[int(x) for x in input().split()]
print(bbsort(a))
```

```python
#4
def nn(a):
    arr=a
    count = {}
    max_count = 0
    result = None

    for x in arr:
        count[x] = count.get(x, 0) + 1

        # chỉ cập nhật khi > max_count
        if count[x] > max_count:
            max_count = count[x]
            result = x

    return str(result) +" xuat hien nhieu nhat + som nhat, " + str(max_count) +" lan"
a=[int(x) for x in input().split()]
print(nn(a))
```

```python
#7
def LIS(a):
    n = len(a)
    dp = [1] * n
    for i in range(n):
        for j in range(i):
            if a[j] < a[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
a = [int(x) for x in input().split()]
print(LIS(a))




```
