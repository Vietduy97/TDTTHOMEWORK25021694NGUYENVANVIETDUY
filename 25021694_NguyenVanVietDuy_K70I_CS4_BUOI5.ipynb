```python
#1
def Max_Of_Two(a,b):
    return max(a,b)
a,b=map(int,input().split())
print(Max_Of_Two(a,b))
```

```python
#2
def swap(a,b):
    a , b=b , a
    return [a,b]
a,b=map(int,input().split())
print(swap(a,b))
```

```python
#3
def snt(n):
    if n<2: return False
    for i in range(2,(int(n**(1/2)))+1):
        if n%i==0:
            return False
    else:
        return True
n=int(input())
print(snt(n))
```

```python
#4
def shh(n):
    t=1
    for i in range(2,n):
        if n%i==0:
            t+=i
    if n==t:
        return True
    else:
        return False
n=int(input())
print(shh(n))
```

```python
#5
def vitri(a,n):
    return a.index(n)
n=int(input())
a=[int(x) for x in input().split()]
print(vitri(a,n))
```

```python
#6
def gt(n):
    t=1
    for i in range(1,n+1):
        t=t*i
    return t

n=int(input())
print(gt(n))

```

```python
#7
def maytinh(s):
    a=s.split()
    if a[1]=='+':
        return float(a[0])+float(a[2])
    if a[1]=='-':
        return float(a[0])-float(a[2])
    if a[1]=='*':
        return float(a[0])*float(a[2])
    if a[1]=='/':
        return float(a[0])/float(a[2])
s=input()
print(round(maytinh(s),2))
```

```python
#8
def kcham(a, b):
    s1 = bin(a)[2:]
    s2 = bin(b)[2:]
    m = max(len(s1), len(s2))
    s1 = s1.zfill(m)
    s2 = s2.zfill(m)
    d = 0
    for i in range(m):
        if s1[i] != s2[i]:
            d += 1
    return d

a, b = map(int, input().split())
print(kcham(a, b))

```

```python
#9
def tongcs(n):
    t=0
    while n>0:
        t=t+n%10
        n=n//10
    return t
n=int(input())
print(tongcs(n))
```

```python
#10
def check(s1,s2)-> bool:
    k=1
    g=1
    ck1=''
    ck2=''
    for i in range(0,len(s1)):
        if s1[i] not in ck1:
            s1=s1.replace(s1[i],str(k))
            k=k+1
            ck1=ck1+s1[i]
        if s2[i] not in ck2:
            s2=s2.replace(s2[i],str(g))
            g=g+1
            ck2=ck2+s2[i]
    if s1==s2:
        return True
    else:
        return False
s1,s2= map(str,input().split())
print(check(s1,s2))
```
