```python
#3
def shp(n):
    s=str(n)
    t=0
    for i in s:
        t=t+(int(i))**2
    return t
n=int(input())
a=[]
try:
    while True:
        if shp(n) in a:
            raise ValueError
        else:
            a.append(shp(n))
        if shp(n)==1:
            print("True")
            break
        n=shp(n)
except ValueError:
    print("False")
```

```python
#10
try:
    n = int(input())
    if n <= 0:
        print("khong hop le")
    else:
        arr = list(map(int, input().split()))
        if len(arr) != n:
            print("khong hop le")
        else:
            ok = True
            for x in arr:
                if x <= 0:
                    ok = False

            if not ok:
                print("khong hop le")
            else:
                if len(arr) != len(set(arr)):
                    print("khong hop le")
                else:
                    print(sum(arr))
except:
    print("khong hop le")
```
