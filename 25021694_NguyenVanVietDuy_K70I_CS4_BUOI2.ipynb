```python
n=int(input())
def sangchu(so):
    match so:
        case 0:
            return "không"
        case 1:
            return "một"
        case 2:
            return "hai"
        case 3:
            return "ba"
        case 4:
            return "bốn"
        case 5:
            return "năm"
        case 6:
            return "sáu"
        case 7:
            return "bảy"
        case 8:
            return "tám"
        case 9:
            return "chín"
print(sangchu(n))
```
