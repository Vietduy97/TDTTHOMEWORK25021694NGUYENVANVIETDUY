```python
def fibo(x):
    a=[0,1,1]
    while a[len(a)-1]+a[len(a)-2]<x:
        a.append(a[len(a)-1]+a[len(a)-2])
    return a[len(a)-1]
n=int(input())
print(fibo(n))


```

```python
so = '0123456789'
kitu='~`!@#$%^&*()_-+={}][|\":<>,.?/'
n = input()
if len(n)<=6:
    print('yeu')
else:
    hoa=0
    kituc=0
    thuong=0
    soc=0
    for i in n:
        if i.isalpha() and i==i.lower():
            thuong=1
        if i.isalpha() and i==i.upper():
            hoa=1
        if i in so:
            soc=1
        if i in kitu:
            kituc=1
    if hoa==1==kituc==thuong==soc:
        print('manh')
    else:
        print('yeu')
```
