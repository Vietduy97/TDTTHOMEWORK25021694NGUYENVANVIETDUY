```python
#3
def sx(n):
    k=n
    g=0
    while True:
        vt=-1
        m=10
        for i in range(len(n)):
            if int(n[i])<=m:
                m=int(n[i])
                vt=i
        if vt==0 and len(n)==2:
            return int(k[:len(k)-2]+k[-1]+k[-2])
        if vt==0:
            n=n[1:]
            g+=1
        if vt!=0:
            vt=vt+g
            break
    i=0
    while int(k[i])==m:
        i=i+1
    return k[:i]+k[vt] + k[i+1:vt] + k[i] + k[vt+1:]
n=input()
print(int(sx(n)))
```

```python
#6
n = int(input())
a = list(map(int, input().split()))
i = n - 2
while i >= 0 and a[i] > a[i + 1]:
    i -= 1
if i < 0:
    a.reverse()
else:
    j = n - 1
    while a[j] < a[i]:
        j -= 1
    a[i], a[j] = a[j], a[i]
    a[i + 1:] = reversed(a[i + 1:])
print(a)

```

```python
#10
def findRadius(houses, heaters):
    if not houses:
        return 0
    if not heaters:
        return 10**9

    houses.sort()
    heaters.sort()

    j = 0
    res = 0

    for h in houses:
        while j + 1 < len(heaters) and \
              abs(heaters[j + 1] - h) <= abs(heaters[j] - h):
            j += 1

        res = max(res, abs(heaters[j] - h))

    return res
nha=[int(x) for x in input().split()]
lo=[int(x) for x in input().split()]
print(findRadius(nha,lo))

```
